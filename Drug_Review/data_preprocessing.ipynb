```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from afinn import Afinn
import pandas as pd
```

```python
#create sentiment score columns using vader and afinn
vader = SentimentIntensityAnalyzer()
afinn = Afinn()

train = pd.read_csv("drugsComTrain_raw.csv")
test = pd.read_csv("drugsComTest_raw.csv")

train["vader"] = train["review"].apply(lambda x: vader.polarity_scores(str(x))["compound"])
train["afinn"] = train["review"].apply(lambda x: afinn.score(str(x)))
test["vader"] = test["review"].apply(lambda x: vader.polarity_scores(str(x))["compound"])
test["afinn"] = test["review"].apply(lambda x: afinn.score(str(x)))

train.to_csv('train_with_sent.csv')
test.to_csv('train_with_sent.csv')
```
